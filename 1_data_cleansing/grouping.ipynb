### Grouping txt files and merge them

```python
import os
from glob import glob
```

```python
input_dir = 'icle_texts'
output_dir = 'icle_grouped'

if not os.path.exists(output_dir):
    os.makedirs(output_dir)
```

```python
# grouping txt files
groups_dict = {}

for path in glob(os.path.join(input_dir, '*.txt')):
    filename = os.path.basename(path)
    group = filename[:4] # first 4 letters = institution that collected data

    if group in groups_dict:
        groups_dict[group].append(filename)
    else:
        groups_dict[group] = [filename]
```

```python
print(groups_dict)
```

```python
# merge txt files in each group (for easier POS taggings)
for group, filenames in groups_dict.items():

    merged_texts = ''

    for file in filenames:
        path = os.path.join(input_dir, file)
        with open(path, 'r', encoding = 'utf-8') as f:
            merged_texts += f.read()
            merged_texts += '\n'
    
    path = os.path.join(output_dir, (group + '.txt'))
    with open(path, 'w', encoding = 'utf-8') as f:
        f.write(merged_texts)
```

After that, I utilized the 'Free CLAWS web tagger' to obatian POS information. This is only available on the web interface, so I had to do that manually for each txt file.

Free CLAWS web tagger: https://ucrel-api.lancaster.ac.uk/claws/free.html

Setting: C5 tagset, vertical output
