```python
import os
from glob import glob
```

```python
def count_lines(input_dir, encoding = 'utf-8'):

    number_of_lines = {}

    for path in glob(os.path.join(input_dir, '*.txt')):
        filename = os.path.basename(path)
        with open(path, 'r', encoding = encoding) as f:
            texts = f.readlines()
        number_of_lines[filename] = len(texts)
    
    sum = 0
    print('Word counting (including metadata lines and punctuations)')
    for filename, number_of_line in number_of_lines.items():
        sum += number_of_line
        print(f'{filename}:\t{number_of_line:,}')
    print(f'Sum:\t{sum:,}')
```

```python
count_lines('icle_lemma_pos', 'utf-8')
```

```python
count_lines('coca_sample', 'latin1')
```
